```python
# Meine Vorhersage: ...
x = 180
print(x > 150, x == 180, x != 180, x <= 179)
print(150 <= x <= 210, 150 < 150 < 210)
print(type(x > 150))
print("180" == x)
```

### 1.2 Vorhersagen: and, or, not
`and` ist nur wahr, wenn **beide** Seiten wahr sind, `or`, wenn **mindestens eine** wahr ist, `not` dreht um. Die Reihenfolge ist `not`, dann `and`, dann `or`.

```python
# Meine Vorhersage: ...
a = True
b = False
print(a and b, a or b, not a, not b)
print(a or b and not a)
print((a or b) and not a)
print(not (a and b), (not a) or (not b))
```

### 2.4 Fehlersuche: Die Bedingung der KI
Die KI hat die Freigabe so geschrieben. Sie läuft, aber sie ist **falsch**: Es sind zwei Fehler drin (einer bei der Verknüpfung, einer bei den Grenzen). Korrigiere die Zeile `freigabe = ...`. Die Prüfzelle rechnet die richtige Antwort selbst aus und gilt für **beliebige** Werte: Ändere die Variablen darüber auf mehrere Szenarien (ein Taster fehlt, Tür offen, Druck 150, 210, 210.1) und prüfe jedes Mal.

```python
taster_links = True
taster_rechts = False
tuer_zu = True
druck = 150.0

freigabe = taster_links or taster_rechts and tuer_zu and 150 < druck < 210
print(freigabe)
```

```python
soll = taster_links and taster_rechts and tuer_zu and 150 <= druck <= 210
pruefe(freigabe, soll, "Freigabe stimmt mit der Vorgabe überein")
```

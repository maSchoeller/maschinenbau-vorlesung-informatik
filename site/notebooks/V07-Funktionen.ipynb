**Kleines Vokabular:** Eine **Funktion** ist ein benannter Baustein (`def name(parameter):`). Ein **Parameter** steht in der Klammer bei `def`, ein **Argument** ist der Wert beim Aufruf. `return` gibt einen **Rückgabewert** zurück und beendet die Funktion. Ohne `return` liefert sie `None`. Ein **Default-Wert** (`tau_zul=40.0`) gilt, wenn das Argument fehlt. Der **Docstring** (`"""..."""`) beschreibt die Funktion mit Einheiten. **Lokale Variablen** gibt es nur in der Funktion.

**Wichtig:** Funktionen **rechnen und geben zurück** (`return`), das **Hauptprogramm gibt aus** (`print`). Wer `print` statt `return` benutzt, bekommt `None`.

**Arbeitsweise:** Erst die Vorhersage-Zellen ausfüllen, dann ausführen. Jede Aufgabe endet mit Selbsttests (`teste`, `teste_nah`): Sie rufen deine Funktion mit festen Werten auf. Läuft sie, steht „OK“, sonst „FEHLER“ mit dem erwarteten und dem tatsächlichen Wert. Gegenprobe von Hand bleibt trotzdem Pflicht: Rechne mindestens einen Wert selbst nach.

```python
# Werkzeug-Zelle für die Selbsttests: einmal ausführen, nicht verändern.
# Jetzt verstehst du, wie sie funktioniert: Es sind zwei eigene Funktionen (def) mit Parametern.
# Sie vergleichen den Wert, den DEINE Funktion geliefert hat (ist), mit dem erwarteten (soll)
# und melden das Ergebnis mit print.
def teste(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def teste_nah(ist, soll, text, toleranz=1e-6):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")
```

### 2.5 Fehlersuche: drei kaputte KI-Funktionen
Eine KI hat drei Funktionen geliefert. Jede hat einen **Fehler**, der **keine Fehlermeldung** auslöst, aber falsch rechnet oder `None` liefert. Finde ihn mit den Selbsttests und repariere die Funktion. Hinweise: Einheit (kW zu W), `print` statt `return`, ein Weg ohne `return`.

```python
import math

def leistung(M, n):
    """Leistung in kW aus Moment M in N·m und Drehzahl n in 1/min."""
    omega = 2 * math.pi * n / 60
    return M * omega

def flaeche(d):
    """Kreisfläche in mm² aus Durchmesser d in mm."""
    print(math.pi * d ** 2 / 4)

def klasse(d):
    """Schruppen (ab 10 mm), Schlichten (ab 1 mm), sonst Feinbearbeitung."""
    if d >= 10:
        return "Schruppen"
    elif d >= 1:
        return "Schlichten"
```

```python
teste_nah(leistung(98.7858, 1450), 15.0, "Leistung aus 98.79 N·m bei 1450 1/min", 0.001)
teste_nah(flaeche(10), 78.53981633974483, "Kreisfläche d=10")
teste(klasse(12), "Schruppen", "klasse(12)")
teste(klasse(5), "Schlichten", "klasse(5)")
teste(klasse(0.5), "Feinbearbeitung", "klasse(0.5)")
```

**Kleines Vokabular:** Ein **Token** ist ein Textstück, mit dem ein Sprachmodell rechnet (hier: ein klein geschriebenes Wort oder der Punkt). Ein **Tokenizer** zerlegt Text in Tokens. **Training** heißt hier: zählen, welches Wort auf welches folgte. Die **Wahrscheinlichkeit** eines Nachfolgers ist sein Anteil an allen Nachfolgern. Die **Temperatur** *T* verändert die Wahrscheinlichkeiten vor dem Würfeln (`p ** (1 / T)`, danach normieren). Ein **Seed** macht den Zufall wiederholbar. Eine **Halluzination** ist eine flüssige, aber nicht belegte oder falsche Aussage eines Sprachmodells.

**So arbeitest du:** Erst jede Vorhersage-Zelle als Kommentar ausfüllen, dann ausführen. Bei den „Selbst bauen“-Aufgaben ersetzt du die `None`-Werte und `TODO`-Zeilen; die Zelle darunter prüft dein Ergebnis und meldet `OK` oder `FEHLER`. Lösungen gibt es hier nicht: Wenn du feststeckst, hilft die Syntax-Karte im Labor oder die Lerneinheit. Phase 2: Du darfst dir von einer KI Vorschläge machen lassen, aber nur, wenn du die Testfälle vorher von Hand aufgeschrieben hast und jeden Vorschlag Zeile für Zeile prüfst.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Wie das innen funktioniert, lernst du in V07. Du musst es nicht verstehen.)
def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")
```

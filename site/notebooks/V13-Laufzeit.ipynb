Die **Laufzeit** ist die Zeit, die ein Programm braucht. Die **Schrittzahl** gibt an, wie viele Rechenschritte es dafür macht. **Median** und **Minimum** sind robuste Kennwerte einer Messreihe.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Es sind nur Funktionen mit def und if aus V04 und V07. Du musst sie nicht ändern.)
import time
import numpy as np

def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float, np.floating)) and abs(float(ist) - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_array(ist, soll, text, toleranz=1e-9):
    # Für Arrays: gleiche Länge und alle Werte nahe beieinander
    if ist is None or type(ist).__name__ != "ndarray":
        print("FEHLER: " + text + " (erwartet wird ein NumPy-Array, bei dir steht " + str(ist) + ")")
    elif len(ist) == len(soll) and all(abs(float(a) - float(b)) <= toleranz for a, b in zip(ist, soll)):
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(list(soll)) + ", bei dir steht " + str(list(ist)) + ")")
```

```python
def zu_dicht(punkte, abstand):
    anzahl = 0
    for i in range(len(punkte)):
        for j in range(i + 1, len(punkte)):
            dx = punkte[i][0] - punkte[j][0]
            dy = punkte[i][1] - punkte[j][1]
            if (dx * dx + dy * dy) ** 0.5 < abstand:
                anzahl = anzahl + 1
    return anzahl

def raster(n):
    # n Punkte auf einem Gitter im Abstand 10 mm: zu_dicht(..., 5) ist immer 0
    return [((k % 40) * 10.0, (k // 40) * 10.0) for k in range(n)]

verhaeltnis = None   # Zeit für 800 Punkte durch Zeit für 400 Punkte
paare_400 = None
paare_800 = None
print(verhaeltnis, paare_400, paare_800)
```

### 2.3 Ein Statuswort lesen (Bitoperatoren, nur lesen)
Ein Antrieb meldet ein 16-Bit-**Statuswort**: jedes Bit ist eine Ja/Nein-Meldung, wie eine Reihe Kontrollleuchten. Bit 0 = bereit, Bit 1 = Freigabe, Bit 3 = Fehler. Mit dem Bitoperator `&` und einer Maske fragst du ein Bit ab: Ist das Ergebnis ungleich 0, ist das Bit gesetzt. Bitoperatoren musst du im Kurs nur lesen können, hier darfst du sie ausprobieren.

**Vorhersage:** Welche der drei Ausgaben sind 0?

```python
status = 0b0000000000001011
print(status & 0b0001)    # Bit 0 (bereit)
print(status & 0b0100)    # Bit 2
print(status & 0b1000)    # Bit 3 (Fehler)
print(bin(status))
```

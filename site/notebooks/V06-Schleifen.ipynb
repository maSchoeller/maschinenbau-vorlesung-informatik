**Kleines Vokabular:** Eine **Schleife** wiederholt einen Block. Ein einmaliges Abarbeiten des Blocks heißt **Durchlauf**. Der **Rumpf** ist der eingerückte Block unter `while` oder `for`. Ein **Zähler** zählt die Durchläufe (`zaehler = zaehler + 1`), ein **Akkumulator** sammelt etwas auf (`summe = summe + wert`; kürzer geht es mit `summe += wert`, das ist dasselbe). Die **Abbruchbedingung** entscheidet, wann die Schleife endet. Wird sie nie falsch, ist es eine **Endlosschleife**.

**Wichtig bei Endlosschleifen:** Hängt eine Zelle fest (der Stern `[*]` bleibt stehen), stoppst du sie in Jupyter mit dem Quadrat-Knopf („Interrupt“) oder im Menü „Kernel → Interrupt“. Das ist dein eigener Watchdog. In dieser Datei läuft keine Zelle mit Absicht endlos, aber ein Tippfehler in deiner Schleife kann es ergeben.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Wie das innen funktioniert, lernst du in V07. Du musst es nicht verstehen.)
def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")
```

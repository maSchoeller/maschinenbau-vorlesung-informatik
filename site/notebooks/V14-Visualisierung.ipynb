**Wichtig bei jeder Diagramm-Aufgabe:** Schreibe `fig, ax = plt.subplots()` (oder bei zwei Teilbildern `fig, (oben, unten) = plt.subplots(2, 1, sharex=True)`) und zeichne auf `ax`. Die Prüfzelle liest `ax` aus. `plt.subplots()` gibt zwei Dinge zurück, die du mit **Tupel-Entpacken** auf zwei Namen verteilst.

**Kleines Vokabular:** Ein **Teilbild** (Achsenobjekt, `ax`) ist ein Diagramm in der Abbildung (`fig`). **Datenkoordinaten** sind deine Zahlen, **Pixel** die Bildschirmpunkte; matplotlib rechnet dazwischen um. Mit `bins` legst du die Zahl der Klassen eines Histogramms fest. Eine **Hilfslinie** (`axhline`, `axvline`) markiert einen Soll- oder Grenzwert.

**Datenschutz:** Keine Firmen- oder Kundendaten in Prompts. Hier genügen Übungsdaten.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Wie das innen funktioniert, lernst du in V07 und V12. Du musst es nicht verstehen.)
def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_ja(bedingung, text):
    # Für Diagramme: Die Bedingung wird aus den Objekten (ax.get_xlabel(), ax.lines ...) gebildet.
    if bedingung:
        print("OK: " + text)
    else:
        print("FEHLER: " + text)
```

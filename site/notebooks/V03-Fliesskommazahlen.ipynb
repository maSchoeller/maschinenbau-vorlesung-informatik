**Kleines Vokabular:** Eine **Fließkommazahl** (`float`) ist eine Zahl mit Nachkommastellen, zum Beispiel `2.5`. Python schreibt den Dezimalpunkt, kein Komma. Der **Datentyp** sagt, was ein Wert ist (Ganzzahl, Kommazahl, Text). Ein **Rundungsfehler** ist die winzige Abweichung, weil der Rechner viele Kommazahlen nur näherungsweise speichern kann. Eine **Toleranz** ist die erlaubte Abweichung vom Sollmaß. Ein **bool** ist ein Wahrheitswert: `True` (wahr) oder `False` (falsch). Er entsteht zum Beispiel aus einem Vergleich wie `3 <= 4`.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Wie das innen funktioniert, lernst du in V04 und V07. Du musst es nicht verstehen.)
def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")
```

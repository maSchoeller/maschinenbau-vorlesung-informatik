**Kleines Vokabular:** Eine **Liste** ist eine veränderbare Reihe von Werten unter einem Namen (`[12.004, 12.011]`), ein **Tupel** eine unveränderliche (`(12.000, 12.018)`), ein **Dictionary** eine Tabelle aus Schlüssel und Wert (`{"W-100": (1, 2.40)}`). Der **Index** zählt ab 0, der **negative Index** von hinten (`-1` ist das letzte Element). Ein **Slice** `liste[a:b]` geht bis vor `b`. **Aliasing** heißt: Zwei Namen zeigen auf dieselbe Liste (`b = a`); eine echte **Kopie** ist `a[:]`.

**So arbeitest du:** Erst jede Vorhersage-Zelle als Kommentar ausfüllen, dann ausführen. Bei den „Selbst bauen“-Aufgaben ersetzt du die `None`-Werte und `TODO`-Zeilen; die Zelle darunter prüft dein Ergebnis und meldet `OK` oder `FEHLER`. Lösungen gibt es hier nicht: Wenn du feststeckst, hilft die Syntax-Karte im Labor oder die Lerneinheit. Phase 1: Copilot nur als Erklärer.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Wie das innen funktioniert, lernst du in V07. Du musst es nicht verstehen.)
def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")
```

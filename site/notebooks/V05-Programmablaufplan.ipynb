**Kleines Vokabular:** Ein **Programmablaufplan (PAP)** ist ein Bild für einen Ablauf aus genormten Symbolen und Pfeilen (DIN 66001): Oval für Start und Ende, Rechteck für eine Rechnung, Parallelogramm für Ein- und Ausgabe, Raute für eine Ja/Nein-Frage. Eine **Verzweigung** ist eine solche Raute: Der Rechner nimmt genau einen der beiden Wege „ja“ oder „nein“ (einen **Zweig**). Ein **Pfad** ist ein ganzer Weg von Start bis Ende. Eine **Einrückung** (4 Leerzeichen) zeigt in Python, welche Zeilen zu einem Zweig gehören. Ein **Randfall** ist ein Wert genau an einer Grenze oder außerhalb des Sinnvollen. Ein **Testfall** ist ein fester Satz Eingaben mit der Ausgabe, die du vorher aus dem PAP abgelesen hast.

**Ablauf in jeder Aufgabe:** 1. Vorhersagen (aufschreiben!) → 2. Ausführen → 3. Selbst bauen.

```python
# Werkzeug-Zelle für die Selbsttests: einfach einmal ausführen, nicht verändern.
# (Wie das innen funktioniert, lernst du in V07 (Funktionen). Du musst es nicht verstehen.)
def pruefe(ist, soll, text):
    if ist == soll:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

def pruefe_nah(ist, soll, text, toleranz=1e-9):
    # Für Kommazahlen: "gleich" heißt hier, die Abweichung ist winzig (Rundungsfehler!).
    if isinstance(ist, (int, float)) and abs(ist - soll) <= toleranz:
        print("OK: " + text)
    else:
        print("FEHLER: " + text + " (erwartet " + str(soll) + ", bei dir steht " + str(ist) + ")")

# Referenz für die Werkstoff-Aufgabe: liefert das richtige Ergebnis für beliebige Testwerte.
def referenz_werkstoff(spannung, temperatur):
    if spannung <= 0 or temperatur < -50 or temperatur > 600:
        return "unplausibel"
    if temperatur > 300:
        return "16Mo3" if spannung <= 200 else "Sonder"
    if spannung <= 140:
        return "S235JR"
    if spannung <= 220:
        return "C45E"
    return "42CrMo4"

# Referenz für die Füllstand-Aufgabe (nennt den Zweig, den ein Wert trifft).
def referenz_status(f):
    if f < 0 or f > 100:
        return "unplausibel"
    if f < 20:
        return "Nachfüllen"
    if f <= 80:
        return "Normal"
    return "Überfüllung droht"

# Referenz für das Schaltjahr.
def referenz_schaltjahr(jahr):
    return jahr % 4 == 0 and (jahr % 100 != 0 or jahr % 400 == 0)
```
